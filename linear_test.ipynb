```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
plt.figure(figsize=(8,5))
plt.title('Relação entre valor da conta e valor da gorjeta')
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.title('Relação entre valor da conta e valor da gorjeta')
sns.regplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
X = np.array(tips['total_bill'].to_list())
Y = np.array(tips['tip'].to_list())
```

```python
b0, b1 = linear_models.linear_regression_closed_form(X, Y)
```

```python
# 244
b0, b1
```

```python
Y_hat = b0 + b1 * X
```

```python
plt.scatter(X, Y)
plt.plot(X, Y_hat, color='red')
```

```python
mae = metrics.mean_absolute_error(np.array(Y), np.array(Y_hat))
mse = metrics.mean_squared_error(np.array(Y), np.array(Y_hat))
rmse = metrics.root_mean_squared_error(np.array(Y), np.array(Y_hat))
```

```python
print(f"MAE: {mae:.3f}\nMSE: {mse:.3f}\nRMSE: {rmse:.3f}")
```
